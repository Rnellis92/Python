```python
import random

random_number = random.randint(10, 99)

guess = int(input("Guess the number between 10 and 99: "))

while guess != random_number:

    if guess < random_number:
        print("Little higher")
    else:
        print("Little lower")

    guess = int(input("Try again: "))

print("Correct! Well done!")
```
